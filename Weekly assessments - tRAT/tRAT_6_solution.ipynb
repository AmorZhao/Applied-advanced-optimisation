# Exercise

This exercise cannot be solved using CVXPY, you need to use pen and paper. The exercise is very simple from a computational point of view, but it requires a good understanding of duality. It is strongly advised that you check with your group the answers to each single question before proceeding too much by yourself.

Consider the optimisation problem

$$
\begin{array}{l}
\displaystyle \min_{x,y} & e^{-x}\\
s.t. & \frac{x^2}{y}\le 0\end{array}\tag{1}
$$

with variables $x$ and $y$ and domain $\mathcal{D} = \{(x,y): y >0\}$.



1.   Justify why this is a convex optimisation problem.
2.   Find the feasibility set and the optimal value of the problem.
3.   Give the Lagrangian.
4.   The dual function is (*the derivation from the Lagrangian is skipped because it is challenging*)
$$
\displaystyle g(\lambda) = \left\{\begin{array}{ll}0 & \lambda \ge 0,\\ -\infty & \lambda < 0\end{array}\right.
$$
Write the dual problem.
5.   Compute the dual optimal value and the duality gap.
6.   Explain why the duality gap is not zero.
7.   (*Hard*) In part 4 the dual function is given because its derivation is difficult. Using the Lagrangian obtained in part 3, attempt to prove that the dual function is the one given in part 4. (*Tip: split the analysis in two cases: $\lambda \ge 0$ and $\lambda < 0$*).
8.   (*Hard*) Consider the optimal value $p^*(u)$ of the perturbed problem
$$
\begin{array}{l}
\displaystyle  \min_{x,y} & e^{-x}\\
s.t. & \frac{x^2}{y}\le u\end{array}
$$
Verify that for $u=1$ and $\lambda^*=0$ the global sensitivity inequality
$$
p^*(u) \ge p^*(0) - \lambda^* u
$$
does not hold.

# Solution

1.   $e^{-x}$ is a convex function because it is an exponential function. $\frac{x^2}{y}$ is a convex function because it is a quadratic-over-linear function.
2.   The feasible set is $\{(0,y): y>0\}$. So the optimal value is $p^*=1$.
3.   The Lagrangian is $L(x,y,\lambda)=e^{-x} + \lambda \frac{x^2}{y}$.
4.   The dual problem is
$$
\begin{array}{l}
\displaystyle \max_{\lambda} & 0\\
s.t. & \lambda\ge 0\end{array}
$$
5.   The dual optimal value is $0$ and the optimal dual variable is $\lambda^* \ge 0$. The duality gap is $p^*-d^* = 1$.
6.   Strong duality does not hold because Slater's condition is not satified because the feasible set of problem $(1)$ does not satisfy the nonlinear inequality constraint strictly.
7.   The dual function is
$$
\displaystyle g(\lambda) = \inf_{x,y>0} \left(e^{-x} + \lambda \frac{x^2}{y} \right).
$$
You can intuitively see that when $\lambda \ge 0$, you can make $L=0$ for $x$ large and $y$ *larger*. This may seem confusing under the belief that we have $\frac{x^2}{y}=\frac{\infty}{\infty}$. However, this intuition is wrong because, since we are on the plane, we can go to infinity at different "speeds" on the $x$ and $y$ directions. This can be shown rigorously: when $\lambda \ge 0$, we can construct sequences of $x_i$s and $y_i$s, such that (for instance) $y_i = x_i^3$ for all $i$. As $x_i \to +\infty$, we have $e^{-x_i} \to 0$ and $\frac{x_i^2}{y_i}=\frac{1}{x_i}=0$. This shows that if on the plane you follow (for instance) the curve $y=x^3$ you will obtain $0$ as infimum. For $\lambda < 0$, we can make $\frac{x^2}{y}$ as large as possible with $x \to +\infty$ and $y$ any finite number, obtaining $0-\infty$. Hence,
$$
\displaystyle g(\lambda) = \inf_{x,y>0} \left(e^{-x} + \lambda \frac{x^2}{y} \right) = \left\{\begin{array}{ll}0 & \lambda \ge 0,\\ -\infty & \lambda < 0\end{array}\right.
$$
8.   For $u=1$, the perturbed problem has optimal value $p^*(u) = 0$, for $x^2 \to \infty$, $y \to \infty$ (because of the reason explained in point 7). The global sensitivity inequality then reads
$$
0 \ge 1 - 0 \cdot u
$$
which does not hold.
